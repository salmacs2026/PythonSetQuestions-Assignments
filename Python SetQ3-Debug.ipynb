```python
# 5.What is wrong with this code?

def is_even(n):
    if n % 2 == 1:        # Logic is incorrect, check 2 == 0.
        return True       # Get input from the user
    else:
        return False
```

```python
#Answer:

def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False
print(is_even(8))
print(is_even(1))
```
